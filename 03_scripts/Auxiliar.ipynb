ESTRUCTURA BÁSICA DE UNA CLASE

```python
class Perro:
    # Constructor: inicializa los atributos del objeto
    def __init__(self, nombre, raza):
        self.nombre = nombre  # Atributo
        self.raza = raza      # Atributo

    # Método: define un comportamiento
    def ladrar(self):
        return f"{self.nombre} dice: ¡Guau!"

# 1. Creación de objetos (Instanciación)
mi_perro = Perro("Rex", "Pastor Alemán")
otro_perro = Perro("Luna", "Labrador")

# 2. Accediendo a los atributos
print(mi_perro.nombre)  # Imprime: Rex
print(otro_perro.raza)  # Imprime: Labrador

# 3. Llamando a los métodos
print(mi_perro.ladrar())  # Imprime: Rex dice: ¡Guau!
```
